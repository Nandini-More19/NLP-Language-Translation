```python
model_name = "facebook/nllb-200-distilled-600M"

tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    src_lang="mar_Deva"
)

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)
model.eval()

print("Model Loaded Successfully!")
print("Using device:", device)
```

```python
marathi_text = "मला सफरचंद आवडते."

inputs = tokenizer(
    marathi_text,
    return_tensors="pt",
    truncation=True
).to(device)

with torch.no_grad():
    generated_tokens = model.generate(
        **inputs,
        forced_bos_token_id=tokenizer.convert_tokens_to_ids("zho_Hans"),
        max_new_tokens=100
    )

chinese_text = tokenizer.batch_decode(
    generated_tokens,
    skip_special_tokens=True
)[0]

print("Marathi:", marathi_text)
print("Chinese:", chinese_text)
```

```python
marathi_text = input("मराठी वाक्य लिहा: ")

inputs = tokenizer(
    marathi_text,
    return_tensors="pt",
    truncation=True
).to(device)

with torch.no_grad():
    generated_tokens = model.generate(
        **inputs,
        forced_bos_token_id=tokenizer.convert_tokens_to_ids("zho_Hans"),
        max_new_tokens=100
    )

print("Chinese translation:", tokenizer.batch_decode(
    generated_tokens,
    skip_special_tokens=True
)[0])
```
